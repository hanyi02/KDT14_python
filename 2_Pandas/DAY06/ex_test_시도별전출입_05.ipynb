```python
## 모듈 로딩
import pandas as pd 
```

[2] 데이터 로딩 <hr>

```python
## xlsx ==> DataFrame으로 로딩
df = pd.read_excel(DATA_FILE, sheet_name="데이터")
```

```python
## -------------------------------------------------------------
## 데이터 전처리 진행 체크 
## -------------------------------------------------------------
## 1. 전출지별 컬럼 => 이전값으로 채우면 결측치 처리
## 2. 0번행 의미 없는 행 => 삭제
## 3. 1970년 컬럼 ~ 2017년 컬럼 타입 object => '-' 문제
## -------------------------------------------------------------
```

[3] 기본 전처리<hr>

```python
## 1. 전출지별 컬럼 => 이전값으로 채우면 결측치 처리
df['전출지별'] = df['전출지별'].ffill()
```

```python
## 2. 0번행 의미 없는 행 => 삭제
df.drop(index=0, inplace=True)
```

```python
## 3. 1970년 컬럼 ~ 2017년 컬럼 타입 object => '-' 문제
df = df.mask(df=='-', 0)
df[df.columns[2:]] = df[df.columns[2:]].astype(int)
df.info()
```

---
#### 문제 5) 서울특별시에서 다른 지역으로 전출한 데이터만 추출하세요.
* 서울특별시 ---> 다른 지역 이동
* 전출지별 : 서울특별시
* 전입지별 : 서울특별시가 아닌 지역명
---

```python
## 전출지별 : 서울특벌시 ==> 전입지별 : 다른 지역 
## 전출지별의 NAN값 채우기
seoulDF = df.copy()
seoulDF["전출지별"] = seoulDF["전출지별"].ffill()
seoulDF.tail(10)
```

```python
## 전출지별 : 서울특벌시 ==> 전입지별 : 다른 지역 
## 전출지별의 NAN값 채우기
#seoulDF[ seoulDF["전출지별"]=="서울특별시" ].head(2)
boolSR = seoulDF["전출지별"]=="서울특별시" 
seoulDF = seoulDF[boolSR]
seoulDF.shape, seoulDF.columns
```

```python
## 전출지별 컬럼 동일값으로 삭제
seoulDF = seoulDF.drop(columns='전출지별')
seoulDF.shape
```

```python
## 전입지별이 서울특별시, 전국이 아닌 행 선택
seoulDF[ (seoulDF['전입지별'] != '전국') & (seoulDF['전입지별'] != '서울특별시') ]
```

```python
## 전입지별이 서울특별시, 전국이 아닌 행 선택
seoulDF[~ seoulDF["전입지별"].isin(["서울특별시", "전국"]) ]
```

```python
## 전입지별이 서울특별시, 전국이 아닌 행 선택
seoulDF.query('전입지별!="서울특별시" & 전입지별 !="전국"').reset_index(drop=True)
```
