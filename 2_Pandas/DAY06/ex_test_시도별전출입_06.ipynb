```python
## -------------------------------------------------------------
## 데이터 전처리 진행 체크 
## -------------------------------------------------------------
## 1. 전출지별 컬럼 => 이전값으로 채우면 결측치 처리
## 2. 0번행 의미 없는 행 => 삭제
## 3. 1970년 컬럼 ~ 2017년 컬럼 타입 object => '-' 문제
## -------------------------------------------------------------
```

[3] 기본 전처리<hr>

```python
## 1. 전출지별 컬럼 => 이전값으로 채우면 결측치 처리
df['전출지별'] = df['전출지별'].ffill()
```

```python
## 2. 0번행 의미 없는 행 => 삭제
df.drop(index=0, inplace=True)
```

```python
## 3. 1970년 컬럼 ~ 2017년 컬럼 타입 object => '-' 문제
df = df.mask(df=='-', 0)
df[df.columns[2:]] = df[df.columns[2:]].astype(int)
df.info()
```

#### 문제 6) 서울특별시로 전입한 지역의 데이터 추출한 문제입니다

[요구사항]
1. 전출지별이  "부산광역시", "대구광역시", "인천광역시", "경기도"  중 하나인 행만 추출하세요.
2. 전입지별, 1990, 2000, 2010, 2017 컬럼만 출력하세요.
3. 2017년 이동자수 기준으로 내림차순 정렬하세요.

- A value is trying to be set on a copy of a slice from a DataFrame
- See the caveats in the documentation: 
- https://pandas.pydata.org/pandas-docs/stable/user_guide/indexing.html#returning-a-view-versus-a-copy

* View란
    * 메모리 낭비를 줄이기 위해 원본 데이터를 복사하지 않고 일부를 바라보는 방식
    * 하지만 View를 수정하면 원본도 영향을 받을 수 있음

* [원인]
    * 원본 DataFrame에서 일부 행/열을 추출해서 만든 DataFrame에 값을 수정/삭제하려고 할 때 발생
    * inplace=True로 원본에 적용할 경우 
    * 이 DataFrame이 원본의 일부인지 복사본인지 불명확
    * 수정한 내용이 원본에 반영될 수도 있고 안 될 수도 있음!!!
* [해결]
    - 일부 데이터를 추출한 뒤 별도로 작업할 목적이면 .copy()로 독립적인 DF로 작업

```python
# 1. 전출지별이  "부산광역시", "대구광역시", "인천광역시", "경기도"  중 하나인 행만 추출하세요.
## => 전입지별이 서울특별시 인 행 추출
mask = df['전입지별'] == '서울특별시'
inSeoulDF = df[mask]

inSeoulDF.drop(columns='전입지별', inplace=True)
inSeoulDF.info()
```

```python
## 전출지가 "부산광역시", "대구광역시", "인천광역시", "경기도"
mask = inSeoulDF["전출지별"].isin(["부산광역시", "대구광역시", "인천광역시", "경기도"])
inSeoulDF = inSeoulDF[mask]
inSeoulDF
```

```python
# 2. 전입지별, 1990, 2000, 2010, 2017 컬럼만 출력하세요.
inSeoulDF = inSeoulDF[[ "전출지별", "1990", "2000", "2010", "2017"]]
inSeoulDF
```

```python
# 3. 2017년 이동자수 기준으로 내림차순 정렬하세요.
inSeoulDF.sort_values(by="2000", ascending=False)
```
