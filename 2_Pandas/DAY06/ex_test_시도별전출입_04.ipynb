[1] 모듈 및 데이터 설정 <hr>

```python
## 모듈 로딩
import pandas as pd 
```

```python
## 데이터 준비
DATA_FILE = '../DATA/시도별 전출입 인구수.xlsx'
```

[2] 데이터 로딩 <hr>

```python
## xlsx ==> DataFrame으로 로딩
df = pd.read_excel(DATA_FILE, sheet_name="데이터")
```

```python
## 앞부분 5개 행을 출력하세요.
print("[앞부분 5개 행]")
df.head()
```

```python
## 데이터의 행수와 열수를 출력하세요.
print(f"[행/열 개수]  {df.shape}")
print(f"행 : {df.shape[0]}개, 열 : {df.shape[1]}개")
```

```python
## 각 컬럼의 자료형을 확인하세요.
print(f"[컬럼별 자료형]")
df.dtypes
```

```python
## -------------------------------------------------------------
## 데이터 전처리 진행 체크 
## -------------------------------------------------------------
## 1. 전출지별 컬럼 => 이전값으로 채우면 결측치 처리
## 2. 0번행 의미 없는 행 => 삭제
## 3. 1970년 컬럼 ~ 2017년 컬럼 타입 object => '-' 문제
## -------------------------------------------------------------
```

[3] 기본 전처리<hr>

```python
## 1. 전출지별 컬럼 => 이전값으로 채우면 결측치 처리
df['전출지별'] = df['전출지별'].ffill()
```

```python
## 2. 0번행 의미 없는 행 => 삭제
df.drop(index=0, inplace=True)
```

```python
## 3. 1970년 컬럼 ~ 2017년 컬럼 타입 object => '-' 문제
df = df.mask(df=='-', 0)
df[df.columns[2:]] = df[df.columns[2:]].astype(int)
df.info()
```

[요구사항]
1. 전출지별, 전입지별, 1970, 1980, 1990, 2000, 2010, 2017 컬럼만 선택하여 새로운 DataFrame을 만드세요.
2. 앞에서 10개 행만 출력하세요.

```python
## 1. 전출지별, 전입지별, 1970, 1980, 1990, 2000, 2010, 2017 컬럼만 선택하여 새로운 DataFrame을 만드세요.
select_cols = ["전출지별", "전입지별", "1970", "1980", "1990", "2000", "2010", "2017"]
partDF = df[select_cols].reset_index(drop=True)
partDF.head(3)
```

```python
## 2. 앞에서 10개 행만 출력하세요.
print("[앞에서 10개 행]")
display(partDF.iloc[:10])
```

```python
## 3. 전출지별, 전입지별, 2017 컬럼만 출력하세요.
print("[전출지별, 전입지별, 2017 컬럼]")
display(partDF.loc[:, ["전출지별", "전입지별", "2017"]].head())
display( partDF[["전출지별", "전입지별", "2017"]].head() )
```

```python
## 4. 0번 행의 2017년 이동자수 값을 선택하여 출력하세요.
print(f"[0번 행의 2017년 이동자수] { partDF.loc[0, '2017']:,}명")
```
