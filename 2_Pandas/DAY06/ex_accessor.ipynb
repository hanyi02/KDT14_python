#### 접근자 (Accessor) 
- Series의 원소/요소값이 특정 데이터타입인 경우 관련 메서드를 활용할 수 있도록 해주는 요소
- 종류
    * 문자형 타입  : Series.str
    * 날짜시간 타입 : Series.dt
    * 범주형 타입 : Series.cat 

```python
## 모듈 로딩
import pandas as pd 
```

```python
## ------------------------------------------------------
## [1] 문자형 접근자 
## ------------------------------------------------------
## SR 인스턴스 생성
nameSR = pd.Series(['Hong Gil-Dong', 'Kim Na-Na', 'Lee Min-Ho'])
display(nameSR)
```

```python
## 원소를 모두 대문자로 변경하고자 할 경우
print("일반 방법------") 
for idx in range(nameSR.shape[0]):
    print(nameSR[idx].upper() )


print("\nstr 접근자 방법----")
nameSR.str.upper()
```

```python
## 원소내 특정 문자 존재 검사 
print("일반 방법------") 
for idx in range(nameSR.shape[0]):
    print('im' in  nameSR[idx] )


print("\nstr 접근자 방법----")
nameSR.str.contains('IM', case=False)
```

```python
## 원소의 성과 이름 분리 컬럼
print("일반 방법------") 
userDF = pd.DataFrame(columns=['Last', 'First'])
for idx in range(nameSR.shape[0]):
    last, first = nameSR[idx].split()
    userDF.loc[userDF.shape[0]+1]= [last, first]
display(userDF)

print("\nstr 접근자 방법----")
nameSR.str.split(expand=True)
```

```python
## ------------------------------------------------------
## [2] 날짜/시간형 접근자 
## ------------------------------------------------------
dataSR = pd.Series(["2024-01-01", "2024-01-15", "2024-02-5"], dtype="datetime64[ns]" )
dataSR
```

```python
## 요일변환
dataSR.dt.day_name()
```

```python
## 시간 추출, 분 추출
dataSR.dt.hour, dataSR.dt.minute
```

```python
## ------------------------------------------------------
## [3] 카테고리 접근자 
## ------------------------------------------------------
df = pd.DataFrame({
    "이름": ["김민수", "이서연", "박지훈", "최유진", "정하늘"],
    "등급": ["A", "B", "A", "C", "B"],
    "지역": ["서울", "부산", "서울", "대구", "부산"]
})

display(df)
print(df.info())
```

```python
## category 타입으로 변환
df["등급"] = df["등급"].astype("category")

print(df.info())
```

```python
## -------------------------
## cat 접근자 활용
## -------------------------
print("카테고리 목록 확인 :",   df["등급"].cat.categories)
print("카테고리 개수 확인 :",   df["등급"].cat.categories.size)
print("카테고리 코드 번호 :\n", df["등급"].cat.codes, sep='')
```

```python
## => 카테고리 이름 변경
df["등급"] = df["등급"].cat.rename_categories({ "A": "우수", "B": "보통", "C": "미흡" })

print("카테고리 이름 변경 :\n", df["등급"], sep='')
```

```python
## => 새 카테고리 추가
df["등급"] = df["등급"].cat.add_categories(["매우우수"])

print("새 카테고리 추가 :\n", df["등급"], sep='')
```
