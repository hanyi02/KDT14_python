### [ Image Read& write ] <hr>
- 경로 지정된 이미지 읽기/ 쓰기, 저장
- 읽기/ 로딩: cv2.imread(경로+ 파일명, 플래그)
- 쓰기/ 저장: cv2.imwrite(저장 경로+ 파일명, 이미지 데이터)

```python
# --------------------------------------------
# 모듈 로딩
# --------------------------------------------
import cv2
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import matplotlib.pyplot as plt
import koreanize_matplotlib
```

```python
# --------------------------------------------
# 이미지 로딩
# --------------------------------------------

IMG_FILE='../Data/Images/img2.png'
colorImg= cv2.imread(IMG_FILE)
```

```python
# --------------------------------------------
# 이미지 데이터 로딩 및 정보 확인
# - 함수: cv2.imwrite(저장 경로+ 파일명, 이미지 데이터)
# --------------------------------------------

SAVE_FILE= './img2.png'

ret= cv2.imwrite(SAVE_FILE, colorImg)

print(ret)
```

```python
# --------------------------------------------
# [실습] S/s 키 입력 시 grayscale 이미지로 저장하기
#       X/x 키 입력 시 종료하기
# --------------------------------------------


IMG_FILE = './img2.png'
img = cv2.imread(IMG_FILE)

if img is None:
    print('이미지를 불러오지 못했습니다.')

else:
    cv2.imshow("image", img)

    key = cv2.waitKey(0) & 0xFF

    if key in [ord('S'), ord('s')]:
        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        cv2.imwrite('gray_img2.png', gray)
        print('grayscale 이미지로 저장했습니다.')

    elif key in [ord('X'), ord('x')]:
        print('종료합니다.')

    cv2.destroyAllWindows()
```

```python
# --------------------------------------------
# [실습] S/s 키 입력 시 grayscale 이미지로 저장하기
#       X/x 키 입력 시 종료하기
# --------------------------------------------


IMG_FILE = './img2.png'
img = cv2.imread(IMG_FILE)


if img is None:
    print('이미지를 불러오지 못했습니다.')
else:
    cv2.imshow("image", img)
    key = cv2.waitKey(0)

    if key in [ord('S'), ord('s')]:
        gray= cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
        cv2.imwrite("gray_ggobuk.png", gray)
    elif key in [ord('X'), ord('x')]:
        print("종료")
    cv2.destroyAllWindows()




```

```python
img=cv2.imread(IMG_FILE)
plt.imshow(img)
plt.show()

bgra_img = cv2.cvtColor(img, cv2.COLOR_BGR2BGRA)
plt.imshow(bgra_img)
plt.show()


rgb_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(rgb_img)
plt.title(f'형태:{rgb_img.shape}, 크기: {rgb_img.size}')
plt.show()

hsv_img1 = cv2.cvtColor(rgb_img, cv2.COLOR_RGB2HSV)
plt.imshow(hsv_img1)
plt.show()

hsv_img1 = cv2.cvtColor(rgb_img, cv2.COLOR_BGR2HSV)
plt.imshow(hsv_img1)
plt.show()
```

```python
# --------------------------------------------
# 이미지 데이터 로딩 및 정보 확인 -(2) 흑백이미지로 읽기
# - 함수: cv2.imread()
# - 옵션/ 플래그: cv2.IMREAD_GRAYSCALE
# --------------------------------------------
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 이렇게만 하면 그레이로 매핑이 안 됨
img= cv2.imread(IMG_FILE, cv2.IMREAD_GRAYSCALE)
plt.imshow(img)
plt.show()

# cmap으로 따로 설정해줘야 그레이이미지로 보임
img= cv2.imread(IMG_FILE, cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap= cm.gray)

plt.title(f'{img.shape}, {img.dtype}, {img.size} 하하')
plt.show()

print(f'{img.shape}, {img.dtype}, {img.size}')

```

```python
# -------------------------------------------
# 이미지 담은 창 띄우기: 특정 키 입력 시 닫기
# -------------------------------------------
# q키 입력 시 닫기
import cv2

IMG_FILE = './img2.png'

img = cv2.imread(IMG_FILE)


if img is None:
    print('이미지를 불러오지 못했습니다.')
else:
    cv2.imshow("image", img)
    
    while True:
        keycode= cv2.waitKey()
        print(f'keycode=> {keycode}')

        if keycode in [ord('q'), ord('Q')]:
            cv2.destroyWindow("image")
            break

```
