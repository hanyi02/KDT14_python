```python
import cv2
import numpy as np

pts1 = np.array([[100, 500], [300, 500], [200, 600]])
pts2 = np.array([[600, 500], [800, 500], [700, 600]])

# 흰색 컬러 캔버스 생성
canvas = np.full((800, 1000, 3), 255, dtype=np.uint8)


## ========================================
## 사각형 그리기: 좌표 두 개 필요 => 너비, 높이 자동 계산
## ========================================

# cv2.rectangle(이미지, 왼쪽위좌표, 오른쪽아래좌표, 색상, 두께)
cv2.rectangle(canvas, (100, 100), (400, 300), (255, 0, 0), 3)

# 내부 채운 사각형
cv2.rectangle(canvas, (500, 100), (800, 300), (0, 255, 0), -1)


## ========================================
## 다각형 그리기: 좌표 여러 개 필요 => 첫 번째와 마지막 좌표 연결
## ========================================

# cv2.polylines(이미지, [좌표배열], 닫힘여부, 색상, 두께)
cv2.polylines(canvas, [pts1], True, (0, 255, 255), 2)
cv2.polylines(canvas, [pts2], True, (255, 0, 255), 2)

# 내부 채운 다각형
pts3 = np.array([[350, 500], [500, 450], [650, 550], [550, 700], [400, 650]])
cv2.fillPoly(canvas, [pts3], (0, 200, 100))


## ========================================
## 원 그리기: 중심점, 반지름 지정
## ========================================

# cv2.circle(이미지, 중심좌표, 반지름, 색상, 두께)
cv2.circle(canvas, (200, 900), 100, (0, 0, 255), 3)

# 내부 채운 원
cv2.circle(canvas, (500, 900), 100, (255, 100, 0), -1)


## ========================================
## 타원/호 그리기: 중심점, 반지름, 회전각, 시작 끝 각도 지정
## ========================================

# cv2.ellipse(이미지, 중심좌표, 축반지름, 회전각, 시작각도, 끝각도, 색상, 두께)

# 전체 타원
cv2.ellipse(canvas, (900, 900), (200, 100), 0, 0, 360, (0, 0, 0), 3)

# 회전된 타원
cv2.ellipse(canvas, (900, 1150), (200, 100), 45, 0, 360, (0, 100, 255), 3)

# 호
cv2.ellipse(canvas, (400, 1200), (200, 100), 0, 0, 180, (255, 0, 0), 5)


cv2.imshow("canvas", canvas)
cv2.waitKey(8000)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

pts1 = np.array([[100, 500], [300, 500], [200, 600]])
pts2 = np.array([[600, 500], [800, 500], [700, 600]])

# 흰색 컬러 캔버스
canvas = np.full((800, 1200, 3), 255, dtype=np.uint8)


## ========================================
## 사각형 그리기
## ========================================

rect_img = canvas.copy()

cv2.rectangle(rect_img, (100, 100), (400, 300), (255, 0, 0), 3)
cv2.rectangle(rect_img, (500, 100), (800, 300), (0, 255, 0), -1)

plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(rect_img, cv2.COLOR_BGR2RGB))
plt.title("Rectangle")
plt.axis("off")
plt.show()


## ========================================
## 다각형 그리기
## ========================================

poly_img = canvas.copy()

cv2.polylines(poly_img, [pts1], True, (0, 255, 255), 2)
cv2.polylines(poly_img, [pts2], True, (255, 0, 255), 2)

pts3 = np.array([[350, 500], [500, 450], [650, 550], [550, 700], [400, 650]])
cv2.fillPoly(poly_img, [pts3], (0, 200, 100))

plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(poly_img, cv2.COLOR_BGR2RGB))
plt.title("Polygon")
plt.axis("off")
plt.show()

## ========================================
## 원 그리기
## ========================================

circle_img = canvas.copy()

cv2.circle(circle_img, (200, 400), 100, (0, 0, 255), 3)
cv2.circle(circle_img, (500, 400), 100, (255, 100, 0), -1)

plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(circle_img, cv2.COLOR_BGR2RGB))
plt.title("Circle")
plt.axis("off")
plt.show()


## ========================================
## 타원/호 그리기
## ========================================

ellipse_img = canvas.copy()

cv2.ellipse(ellipse_img, (300, 300), (200, 100), 0, 0, 360, (0, 0, 0), 3)
cv2.ellipse(ellipse_img, (800, 300), (200, 100), 45, 0, 360, (0, 100, 255), 3)
cv2.ellipse(ellipse_img, (600, 600), (200, 100), 0, 0, 180, (255, 0, 0), 5)

plt.figure(figsize=(8, 8))
plt.imshow(cv2.cvtColor(ellipse_img, cv2.COLOR_BGR2RGB))
plt.title("Ellipse / Arc")
plt.axis("off")
plt.show()
```
