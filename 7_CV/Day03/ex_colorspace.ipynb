```python
## 모듈 로딩
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
```

```python
## 이미지 로딩
IMG_FILE='../Images/opencv_logo.png'
imgNP3= cv2.imread(IMG_FILE)
imgNP4= cv2.imread(IMG_FILE, cv2.IMREAD_UNCHANGED)
```

```python
print('------- imgNP3 정보')
print(f'{imgNP3.shape}, {imgNP3.ndim}')


print('------- imgNP4 정보')
print(f'{imgNP4.shape}, {imgNP4.ndim}')
```

```python
fig, axes= plt.subplots(1, 4, sharey= True)

b, g, r, a= cv2.split(imgNP4)


for ax, data, title in zip(axes, (b, g, r, a ), ('B', 'G', 'R', 'A')):

    ax.set_title(f'{title} Channel')
    ax.imshow(data, cmap=cm.gray)
```
