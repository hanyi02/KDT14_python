- 이미지 데이터 양이 많음=> 메모리/ 연산량(속도) 문제
- 필요한 부분만 추출 필수!=> ROI
- 방법
    * (1) Numpy로 지정된 위치에 슬라이싱=> view type으로 copy() 필요
    * (2) openCV의 cv2.selectROI()/ cv2.selectROIs() => 선택한 부분의 좌표값 반환


```python
## 모듈 로딩
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
```

```python
IMG_FILE='../Images/bts_01.jpg'
SAVE_FILE='./crop'

```

```python
imgNP= cv2.imread(IMG_FILE)

roi= cv2.selectROI("SELECT", imgNP)
plt.imshow( imgNP[roi[1]: roi[1]+roi[3], roi[0]: roi[0]+roi[2] ])
plt.show()

cv2.imwrite('./roi.jpg', imgNP[ roi[1]: roi[1]+roi[3], roi[0]: roi[0]+roi[2] ] )

cv2.waitKey(3000)
cv2.destroyAllWindows()

```
