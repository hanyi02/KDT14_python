```python

# 동영상 또는 연결 카메라
VIDEO = '../Video/person_02.mp4'
CAM_ID = 0
F_NAME = 'person_detect'

os.makedirs(F_NAME, exist_ok=True)

MODEL_FILE = '../Haar/haarcascade_frontalface_alt2.xml'
# MODEL_FILE1 = '../Haar/haarcascade_smile.xml'

face_detector = cv2.CascadeClassifier(MODEL_FILE)
# smile_detector = cv2.CascadeClassifier(MODEL_FILE1)

cap = cv2.VideoCapture(VIDEO)

if cap.isOpened():
    cnt = 0

    while True:
        ret, img = cap.read()

        if not ret:
            break

        gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

        faces = face_detector.detectMultiScale(
            gray,
            scaleFactor=1.1,
            minNeighbors=7,
            minSize=(30, 30)
        )

        print(f'얼굴 감지결과: {len(faces)}개, {faces}')

        for (x, y, w, h) in faces:
            cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)

            face_gray = gray[y:y + h, x:x + w]
            face_color = img[y:y + h, x:x + w]

            faces = face_detector.detectMultiScale(
                face_gray,
                scaleFactor=1.8,
                minNeighbors=20,
                minSize=(20, 20)
            )

        cv2.imshow("person_face_detect", img)
        cv2.imwrite(f'./{F_NAME}/{F_NAME}_{cnt:0>3}.jpg', img)

        cnt += 1

        if cv2.waitKey(25) & 0xFF == 27:
            break

else:
    print(f'{VIDEO}: 연결 또는 초기화 실패')

cap.release()
cv2.destroyAllWindows()
```
