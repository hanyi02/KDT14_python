### [ 컬러 스페이스 - hsv ] <hr>

- HSV 색공간
    * 특징: H 채널이 조명 변화와 무관하게 안정적-> 색깔로 물체 분리에 최적
    * HSI, HSL 이라고 함 (세부 계산 방식은 다름)
    * cv2.cvtColor(img, CV2_COLOR_RGB2HSV)

```python
## 모듈 로딩
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
```

```python
## 이미지 로딩
imgNP= cv2.imread('../Images/cat_03.jpg')
imgHSV= cv2.cvtColor(imgNP, cv2.COLOR_BGR2HSV)
b, g, r= cv2.split(imgNP)
h, s, v= cv2.split(imgHSV)
```

```python
## ------------------------------------------------------------
## 함수이름 : show_channels
## 함수기능 : 한 행에 원본 + 채널들을 표시하는 함수
## ------------------------------------------------------------
def show_channels(axes_row, channels, names, cmaps, original_img):

    # 행에서 1번째 즉, 0번 컬러 : 원본 이미지 출력
    axes_row[0].imshow(cv2.cvtColor(original_img, cv2.COLOR_BGR2RGB))
    axes_row[0].set_title("원본 이미지")

    # 행에서 1번 컬러부터 채널별 출력
    for ax, ch, name, cmap in zip(axes_row[1:], channels, names, cmaps):
        ax.imshow(ch, cmap=cmap)
        ax.set_title(name)
```

```python
## ============================================================
## BGR 채널 vs HSV 채널 비교 그래프
## -> 2행 4열
## -> 행별로 ColorSpace에 따른 색상 구분 가능
## ============================================================
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
fig.suptitle("BGR 채널 vs HSV 채널 비교", fontsize=15, fontweight="bold")

## BGR 원본과 BGR 채널 분리
show_channels(axes[0], [b, g, r], ["B 채널", "G 채널", "R 채널"], ["Blues", "Greens", "Reds"], imgNP)

## BGR 원본과 HSV 채널 분리
show_channels(axes[1], [h, s, v], ["H (색상)", "S (채도)", "V (명도)"], ["hsv", "YlOrRd", "gray"], imgNP)

## 피규어 안에 그래프 맞추기
plt.tight_layout()

## 화면 파일로 저장
plt.savefig("result_hsv_channels.jpg", dpi=120, bbox_inches="tight")
plt.show()

print("결과 저장: result_hsv_channels.jpg")
print("-> H 채널을 보면 색깔별로 픽셀값이 뚜렷하게 구분됨을 확인할 수 있습니다.")
```
