```python
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
```

```python
## 이미지 데이터와 모델 파일

IMG_FILE='../Day03/image.png'
MODEL_FILE='../Haar/haarcascade_frontalface_alt2.xml'
```

```python
MODEL_FILE='../Haar/haarcascade_frontalface_alt2.xml'
detector= cv2.CascadeClassifier(MODEL_FILE)

imgNP= cv2.imread(IMG_FILE)
result= detector.detectMultiScale(imgNP)
print(f'감지결과: {len(result)}개, {result}')
```

```python
x, y, w, h= result[3]
cv2.rectangle(imgNP, (x, y), (x+w, y+h), (255, 0, 0), 3)

cv2.imshow("detect", imgNP)
cv2.waitKey(4000)
cv2.destroyAllWindows()
```
