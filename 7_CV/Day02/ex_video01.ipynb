### [ 동영상/ 카메라 영상 제어 ]  <hr>


- 영상: 이미지를 초당 27~30 개 보여주는 것
- 단위: FPS(frames per second)
- 전용 클래스: VideoCapture 클래스
- 필수 메서드: 
    * 연결 및 초기화 여부 체크: isOpened()-> T/F
    * 프레임 읽기: read()-> T/F, ndarray image data
    * 연결 해제: release()

```python
import cv2
import os


# 동영상 또는 연결 카메라
VIDEO= '../Video/big_buck.avi'
CAM_ID=0 # 컴퓨터와 연결된 카메라 번호
F_NAME= 'big_buck'

os.makedirs(F_NAME, exist_ok=True)
```

```python
## 동영상 또는 카메라 객체 생성
cap= cv2.VideoCapture(VIDEO)

# 객체 초기화 연결 여부
if cap.isOpened():
    cnt=0
    while True:
        ret, img= cap.read()

        if ret:
            cv2.imshow("big_buck", img)
            cv2.imwrite(f'./{F_NAME}/{F_NAME}_{cnt:0>3}.jpg', img)
            cnt+=1
            cv2.waitKey(25)
        else:
            break
else:
    print(f'{VIDEO}: 연결 또는 초기화 실패')


## 동영상 또는 카메라 객체가 사용하는 메모리 해제

cap.release()

## 띄워진 창 모두 닫기
cv2.destroyAllWindows()
```
