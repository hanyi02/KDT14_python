### [ 이미지 크기 변경 ]  <hr>
- 가장 많이/ 필수로 하는 전처리 과정
- 이미지마다 크기가 다르기 때문에 모든 이미지 크기 일치를 위해서 사용됨
- 종류:
    * 절대 크기 변경: 이미지 크기 고려하지 않고 동일 크기 일치
    * 상대 크기 변경: 이미지 크기 고려해서 크기 조정


```python
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
```

```python
## 데이터 준비

FILE_PATH= '../Images/girl.jpg'
```

```python
# 이미지 데이터 로딩
imgNP= cv2.imread(FILE_PATH)

print(f'shape: {imgNP.shape}, nDim: {imgNP.ndim}D')
```

```python
# openCV& matplotlib
# openCV로 이미지 로딩 시 채널 순서가 BGR
# 일반적인 이미지 SW/Package는 채널 순서 RGB
# 다른 곳에 사용 시 채널 순서 변경 필수!

plt.imshow(imgNP)
plt.show()


# cv2.imshow("girl", imgNP)
# cv2.waitKey()
# cv2.destroyWindow("girl")


img1 = cv2.cvtColor(imgNP, cv2.COLOR_RGB2BGR)
plt.imshow(img1)
plt.show()
```

```python
# ===================================
# 이미지 크기 조정
# ====================================
imgRGB= cv2.cvtColor(imgNP, cv2.COLOR_BGR2RGB)
imgGRAY= cv2.cvtColor(imgNP, cv2.COLOR_BGR2GRAY)


imgAbs= cv2.resize(imgRGB, (100, 100))
imgRel= cv2.resize(imgRGB, (0, 0), fx=1.5, fy=0.5)


fig, (ax1, ax2) = plt.subplots(1, 2)

ax1.imshow(imgAbs)
ax1.set_title('abs')
ax2.imshow(imgRel)
ax2.set_title('Rel')

plt.tight_layout()
plt.show()
```

```python
imgAbs= cv2.resize(imgRGB, (100, 100))
imgRel= cv2.resize(imgRGB, (0, 0), fx=1.5, fy=0.5)


plt.imshow(imgRGB)
plt.show()

plt.imshow(imgAbs)
plt.show()

plt.imshow(imgRel)
plt.show()
```

```python

# 샤프닝 마스크 원소 지정 
data1 = [0, -1, 0,
        -1, 5, -1,
         0, -1, 0]
data2 = [[-1, -1, -1],
         [-1, 9, -1],
         [-1, -1, -1]]
mask1 = np.array(data1, np.float32).reshape(3, 3)
mask2 = np.array(data2, np.float32)

sharpen1 = cv2.filter2D(imgAbs, -1, mask1)
sharpen2 = cv2.filter2D(imgAbs, -1, mask2)
sharpen1 = cv2.convertScaleAbs(sharpen1)
sharpen2 = cv2.convertScaleAbs(sharpen2)



plt.imshow(sharpen1)
plt.show()

plt.imshow(sharpen2)
plt.show()
```
