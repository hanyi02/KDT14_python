### [ 동영상/ 카메라 영상 제어 ]  <hr>

- VideoWriter 동영상 저장 클래스
- 필수 매서드
    * fourcc( c1, c2, c3, c4): 포맥 방식 문자 설정 => cv2.VideoWriter_fourcc() 함수
    * write(ndarray 이미지): 비디오 프레임으로 저장
    * release(): VideoWriter 객체 해제

```python
import cv2
import os



```

```python
## -----------------------------------------
## 컴퓨터 연결된 카메라 로딩
## -----------------------------------------

CAMERA_ID=0
cap= cv2.VideoCapture(CAMERA_ID)
print(f'{cap.isOpened()}')


## 현재 정보 출력
print(f'{cap.get(cv2.CAP_PROP_FORMAT)}') ## 포맷
print(f'{cap.get(cv2.CAP_PROP_FOURCC)}') # 4개의 캐릭터
print(f'{cap.get(cv2.CAP_PROP_FPS)}') # 프레임수
print(f'{cap.get(cv2.CAP_PROP_FRAME_HEIGHT)}') # 프레임 높이
print(f'{cap.get(cv2.CAP_PROP_FRAME_WIDTH)}') #프레임 너비
print(f'{cap.get(cv2.CAP_PROP_FRAME_COUNT)}') #프레임 개수

# 현재 속성 변경
cap.set(cv2.CAP_PROP_FRAME_WIDTH, 300)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT, 200)

print(f' 변경 후: {cap.get(cv2.CAP_PROP_FRAME_WIDTH)}')
print(f' 변경 후: {cap.get(cv2.CAP_PROP_FRAME_HEIGHT)}')

```
