```python
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm

```

```python
# ============================================================
# 1. NumPy 배열 생성 (3.1.3 값으로 생성)
# ============================================================

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([
    [1, 2, 3, 4],
    [5, 6, 7, 8]
])
arr3 = np.array([1, 2, 3.14, 4])
arr4 = np.array([1, 2, 3, 4], dtype=np.float64)

print('arr1:', arr1, arr1.dtype, arr1.shape)
print('arr2:\n', arr2, arr2.dtype, arr2.shape)
print('arr3:', arr3, arr3.dtype, arr3.shape)
print('arr4:', arr4, arr4.dtype, arr4.shape)
```

```python
# ============================================================
# 2. arange 생성과 dtype 확인
# ============================================================

arr5 = np.arange(5)
arr6 = np.arange(3.0)

print('arr5:', arr5, arr5.shape, arr5.dtype)
print('arr6:', arr6, arr6.shape, arr6.dtype)
```

```python
# ============================================================
# 3. empty, zeros, ones, full (3.1.4 크기와 초기값으로 생성)
# ============================================================

empty_arr = np.empty((2, 3))
empty_int = np.empty((2, 3), dtype=np.int16)
zeros_arr = np.zeros((2, 3))
ones_arr = np.ones((2, 3), dtype=np.float32)
full_arr = np.full((2, 3, 4), 255, dtype=np.uint8)

print('empty_arr:\n', empty_arr, empty_arr.dtype, empty_arr.shape)
print('empty_int:\n', empty_int, empty_int.dtype, empty_int.shape)
print('zeros_arr:\n', zeros_arr, zeros_arr.dtype, zeros_arr.shape)
print('ones_arr:\n', ones_arr, ones_arr.dtype, ones_arr.shape)
print('full_arr:\n', full_arr, full_arr.dtype, full_arr.shape)
```

```python
# ============================================================
# 4. 이미지와 같은 shape 배열 만들기
# ============================================================

if cv2 is not None:
    img_path = '../Images/cat_00.jpg'
    img = cv2.imread(img_path)

    if img is not None:
        empty_like_img = np.empty_like(img)
        zeros_like_img = np.zeros_like(img)
        ones_like_img = np.ones_like(img)
        full_like_img = np.full_like(img, 255)

        print('img shape:', img.shape, img.dtype)
        print('empty_like_img:', empty_like_img.shape, empty_like_img.dtype)
        print('zeros_like_img:', zeros_like_img.shape, zeros_like_img.dtype)
        print('ones_like_img:', ones_like_img.shape, ones_like_img.dtype)
        print('full_like_img:', full_like_img.shape, full_like_img.dtype)
```

```python
# ============================================================
# 5. dtype 변환
# ============================================================

arr7 = np.arange(10)
arr8 = arr7.astype('float32')
arr9 = np.uint8(arr8)
arr10 = arr9.astype(np.float64)

print('arr7:', arr7, arr7.dtype)
print('arr8:', arr8, arr8.dtype)
print('arr9:', arr9, arr9.dtype)
print('arr10:', arr10, arr10.dtype)
```

```python
# ============================================================
# 6. reshape, ravel, transpose
# ============================================================

arr11 = np.arange(6)
arr12 = arr11.reshape(2, 3)
arr13 = np.arange(24).reshape(2, 3, 4)
arr14 = np.arange(100).reshape(2, -1)
arr15 = np.arange(100).reshape(-1, 5)
arr16 = np.ravel(arr13)
arr17 = np.arange(10).reshape(2, -1)

print('arr11:', arr11, arr11.shape)
print('arr12:\n', arr12, arr12.shape)
print('arr13:\n', arr13, arr13.shape)
print('arr14:\n', arr14, arr14.shape)
print('arr15:\n', arr15, arr15.shape)
print('arr16:', arr16, arr16.shape)
print('arr17.T:\n', arr17.T)
```

```python
# ============================================================
# 7. 인덱싱
# ============================================================

arr18 = np.arange(10)
arr19 = np.arange(12).reshape(4, 3)

print('arr18:', arr18)
print('arr18[5]:', arr18[5])
print('arr18[2:7]:', arr18[2:7])
print('arr18[:5]:', arr18[:5])
print('arr18[5:]:', arr18[5:])
print('arr18[::2]:', arr18[::2])
print('arr18[::-1]:', arr18[::-1])

print('arr19:\n', arr19)
print('arr19[0, 0]:', arr19[0, 0])
print('arr19[1]:', arr19[1])
print('arr19[:, 1]:', arr19[:, 1])
print('arr19[1:3, 1:3]:\n', arr19[1:3, 1:3])
```

```python
# ============================================================
# 8. 리스트 연산과 NumPy 벡터 연산
# ============================================================

mylist = list(range(10))
print('mylist 원본:', mylist)

for i in range(len(mylist)):
    mylist[i] = mylist[i] + 1

print('mylist + 1:', mylist)

arr20 = np.arange(10)
print('arr20 원본:', arr20)
print('arr20 + 1:', arr20 + 1)

```

```python
# ============================================================
# 9. 스칼라 연산과 비교 연산
# ============================================================

arr21 = np.arange(5)
arr22 = arr21+ 5
arr23 = arr21- 2
arr24 = arr21* 2
arr25 = arr21/ 2
arr26 = arr22** 2
arr27 = arr22> 2

print('arr21:', arr21)
print('arr21+ 5:', arr22)
print('arr21- 2:', arr23)
print('arr21* 2:', arr24)
print('arr21/ 2:', arr25)
print('arr22** 2:', arr26)
print('arr22> 2:', arr27)
```

```python
# ============================================================
# 10. 이미지 배열 정보 확인
# ============================================================

if cv2 is not None:
    img_path = '../Images/boy_face.jpg'
    img = cv2.imread(img_path)

    if img is not None:
        print(type(img))
        print('ndim:', img.ndim)
        print('shape:', img.shape)
        print('size:', img.size)
        print('dtype:', img.dtype)
        print('itemsize:', img.itemsize)
```

```python
# ============================================================
# 11. 그레이스케일 이미지 직접 생성
# ============================================================

if cv2 is not None:
    gray_img = np.zeros((120, 120), dtype=np.uint8)

    gray_img[25:35, :] = 45
    gray_img[55:65, :] = 115
    gray_img[85:95, :] = 160
    gray_img[:, 35:45] = 205
    gray_img[:, 75:85] = 255

    cv2.imshow('Gray', gray_img)
    if cv2.waitKey(0) & 0xFF == 27:
        cv2.destroyAllWindows()

```

```python
# ============================================================
# 12. BGR 컬러 이미지 직접 생성
# ============================================================

if cv2 is not None:
    bgr_img = np.zeros((120, 120, 3), dtype=np.uint8)

    bgr_img[25:35, :] = [255, 0, 0]
    bgr_img[55:65, :] = [0, 255, 0]
    bgr_img[85:95, :] = [0, 0, 255]
    bgr_img[:, 35:45] = [255, 255, 0]
    bgr_img[:, 75:85] = [255, 0, 255]

    cv2.imshow('BGR', bgr_img)
    if cv2.waitKey(0) & 0xFF == 27:
        cv2.destroyAllWindows()
```
