### [ 이미지 색공간 변경 ]  <hr>
- 색 공간: 이미지 색상을 표현하는 다양한 방식들. 목적에 따라 적합한 색공간 사용 필요성
- 변환 함수: cv2.cvtColor()
- 색 공간 변환 상수: cv2.COLOR_BGR2RGB, ....


```python
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
```

```python
## 데이터 준비

FILE_PATH= '../Images/girl.jpg'
```

```python
# 이미지 데이터 로딩
imgNP= cv2.imread(FILE_PATH)

print(f'shape: {imgNP.shape}, nDim: {imgNP.ndim}D')
```

```python
# openCV& matplotlib
# openCV로 이미지 로딩 시 채널 순서가 BGR
# 일반적인 이미지 SW/Package는 채널 순서 RGB
# 다른 곳에 사용 시 채널 순서 변경 필수!

plt.imshow(imgNP)
plt.show()


# cv2.imshow("girl", imgNP)
# cv2.waitKey()
# cv2.destroyWindow("girl")


img1 = cv2.cvtColor(imgNP, cv2.COLOR_RGB2BGR)
plt.imshow(img1)
plt.show()
```

```python
# ===================================
# 색 공간 변경: cv2.cvtColor(이미지, cv2.COLOR_@@@@@@@@ )
# ====================================
imgRGB= cv2.cvtColor(imgNP, cv2.COLOR_BGR2RGB)
imgGRAY= cv2.cvtColor(imgNP, cv2.COLOR_BGR2GRAY)


fig, (ax1, ax2, ax3) = plt.subplots(1, 3)

ax1.imshow(imgNP)
ax1.set_title('BGR')
ax2.imshow(imgRGB)
ax2.set_title('RGB')
ax3.imshow(imgGRAY, cmap=cm.gray)
ax3.set_title('Gray')

plt.tight_layout()
plt.show()
```

```python


import cv2
import numpy as np
import matplotlib.pyplot as plt

# 채널 분리
b_channel, g_channel, r_channel = cv2.split(imgNP)

zeros = np.zeros_like(b_channel)

# matplotlib은 RGB 순서로 보여줌
b_img = cv2.merge([zeros, zeros, b_channel])  # 파랑만 보이게
g_img = cv2.merge([zeros, g_channel, zeros])  # 초록만 보이게
r_img = cv2.merge([r_channel, zeros, zeros])  # 빨강만 보이게

# 원본 컬러 이미지로 다시 합치기
# OpenCV 기준: BGR
total_bgr = cv2.merge([b_channel, g_channel, r_channel])

# matplotlib 출력용 RGB 변환
total_rgb = cv2.cvtColor(total_bgr, cv2.COLOR_BGR2RGB)

fig, axes = plt.subplots(1, 4, figsize=(16, 4))

axes[0].imshow(b_img)
axes[0].set_title('B channel color')
axes[0].axis('off')

axes[1].imshow(g_img)
axes[1].set_title('G channel color')
axes[1].axis('off')

axes[2].imshow(r_img)
axes[2].set_title('R channel color')
axes[2].axis('off')

axes[3].imshow(total_rgb)
axes[3].set_title('Total color')
axes[3].axis('off')

plt.tight_layout()
plt.show()

```
