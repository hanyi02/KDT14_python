```python
## -------------------------------------
## 이미지 데이터 로딩 & 기본 정보
## -------------------------------------
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
import sys
sys.path.append(r"C:\KDT14\7_CV")
import cv_utils
IMG_FILE= '../Images/girl.jpg'
imgNP= cv2.imread(IMG_FILE)
```

```python
## ------------------------------------------
## 이미지 연산=> 밝기 조절. 0에 가까운 값- 어둡게, 255에 가까운 값- 밝게
## ------------------------------------------
testNP= imgNP.copy()
testNP= testNP.astype(np.int16)

```

```python
## ------------------------------------
## 덧셈& 뺄셈 연산
## ------------------------------------
## 덧셈=> 밝게
addNPs= [testNP + value for value in range(10, 131, 40)]
titles= [f'+{value}' for value in range(10, 131, 40)]

for arr in addNPs:
    arr[arr>255]= 255

cv_utils.printPlots(1, 5, ['원본', '+10', '+50', '+90', '+120'], [imgNP]+ addNPs)
```
