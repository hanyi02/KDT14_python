### [ 이진화/ 스레시홀딩 ] <hr>

- 이미지의 형태/ 외형/ 윤곽 정보를 추출할 수 있는 방식
- 이미지에서 원하는 부분만 추출할 수 있는 방식
- 픽셀을 0과 1(또는 255, 특정값) 으로 변환하는 것
- 픽셀 값 변환을 위한 기준값. 경계값, 임계값 설정이 중요함

```python
## 모듈 로딩
import cv2
import os
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np
import matplotlib.cm as cm
import sys
sys.path.append(r"C:\KDT14\7_CV")
import cv_utils
```

```python
IMG_FILE= '../Images/gray_gradient.jpg'
imgNP= cv2.imread(IMG_FILE, cv2.IMREAD_GRAYSCALE)
print(f'{imgNP.shape}, {imgNP.ndim}, {imgNP.dtype}')
```

```python
## ---------------------------------------
# 이진화 방식 [1]
## ---------------------------------------

binNP= np.zeros_like(imgNP)
print(f'{binNP.shape}, {binNP.ndim}, {binNP.dtype}')

TH_VALUE= 127

binNP[imgNP>TH_VALUE]=255

ret, thresh_cv= cv2.threshold(imgNP, 127, 255, cv2.THRESH_BINARY)
# _, axes= plt.subplots(1, 2, sharey=True)


imgs= {'Original': imgNP, 'NP': binNP, 'cv2.threshhold': thresh_cv}
for i, (key, value) in enumerate(imgs.items()):
    plt.subplot(1, 3, i+1)
    plt.title(key)
    plt.imshow(value, cmap='gray')
```
