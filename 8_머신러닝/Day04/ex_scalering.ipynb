```python

## -------------------------------------
## 이미지 데이터 로딩 & 기본 정보
## -------------------------------------
import cv2
import os
import pandas as pd
import matplotlib.pyplot as plt
import koreanize_matplotlib
import numpy as np

from sklearn.model_selection import train_test_split #학습용| 테스트용 데이터 분할
from sklearn.model_selection import KFold, StratifiedKFold
from sklearn.svm import SVC

## 모델
from sklearn.neighbors import KNeighborsClassifier 


from sklearn.metrics import accuracy_score, classification_report # 평가
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import cross_validate

## 교차검증+ 튜닝
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV 

## 모듈 로딩
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler
```

```python
## 임시 데이터
dataDF= pd. DataFrame ({

'Height': [158, 192, 172, 164, 155],
'Weight': [50, 81, 88, 41, 52] })

dataDF
```

- **Scikit-learn 스케일러 사용법**
    * 인스턴스 생성
    * fit() 메서드에 데이터 전달=> 스케일링을 위한 데이터 기반 기준값 계산
    * transform() 매서드에 데이터 전달=> 데이터 스케일링 진행
    * fit_transform() 매서드=> fit() 기능+ transform() 기능

```python
## ===================================================
## [1] StandardScaler: 평균 0, 표준편차 1이 되도록 스케일링
## ===================================================
## 인스턴스 생성
stdScaler= StandardScaler()

## 데이터 전달
stdScaler.fit(dataDF)

## 스케일링 진행
scaled_dataNP= stdScaler.transform(dataDF)

print(scaled_dataNP.mean().round(1), scaled_dataNP.std().round(1))

print(scaled_dataNP)
```

```python
## ===================================================
## [2] MinMaxScaler: 0~1 사이로 변환
## ===================================================
## 인스턴스 생성
mmScaler= MinMaxScaler()

## 데이터 전달
mmScaler.fit(dataDF)

## 스케일링 진행
scaled_dataNP= mmScaler.transform(dataDF)

print(scaled_dataNP.mean().round(1), scaled_dataNP.std().round(1))

print(scaled_dataNP)
```
