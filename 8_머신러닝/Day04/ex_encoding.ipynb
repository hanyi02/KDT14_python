```python

## -------------------------------------
## 이미지 데이터 로딩 & 기본 정보
## -------------------------------------
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, OrdinalEncoder
import pandas as pd
```

```python
## 임시 데이터
dataDF= pd.DataFrame({ 'Height': [158, 192, 172, 164, 155],
                        'Weight': [50, 81, 110, 41, 81],
                        'Blood' : ['B', 'O', 'O', 'A', 'AB'],
                        'Grade' : ['첫번째', '두번째', '첫번째', '세번째', '두번째'],
                        'B_TYPE':[ '정상' ,'마름' , '비만' , '정상' , '비만']})

dataDF
```

```python
## Blood, B_TYPE 컬럼 자료형 변환
dataDF['Blood']= dataDF['Blood'].astype('category')
dataDF['B_TYPE']= dataDF['B_TYPE'].astype('category')


print(dataDF.dtypes)
```

- **인코딩**

```python
## ===================================================
## [1] LabelEncoder: 타겟 컬럼위 슈치화
##                   0~ (클래스 -1)
##                   예: iris 3개 품종/ 클래스=> 0~2
##                      .fit(1D)
## ===================================================
## 인스턴스 생성
lblEncoder= LabelEncoder()

## 데이터 전달
lblEncoder.fit(dataDF['B_TYPE'])
print("classses_", lblEncoder.classes_)

## 인코딩 진행
encode_target= lblEncoder.transform(dataDF['B_TYPE'])

print(dataDF['B_TYPE'].to_list())
print(encode_target)

```

```python
## ===================================================
## [2] OneHotEncoder: 텍스트 피처, 순서 없는 텍스트 데이터
## ===================================================
## 인스턴스 생성
ohEncoder=  OneHotEncoder(sparse_output=False)

## 데이터 전달 : 2D
ohEncoder.fit(dataDF[['Blood']])
print(ohEncoder.categories_)
print(ohEncoder.drop_idx_)
print (ohEncoder. feature_names_in_)

encode_DF= ohEncoder.transform(dataDF[['Blood']])

print(dataDF['Blood'].to_list())
print(encode_DF)
```

```python
## ===============================
## [3] OrdinalEncoder : 텍스트 피쳐. 순서 있는 텍스트 데이터
##                      .fit(2D)
## ===============================
odEncoder = OrdinalEncoder()

## 데이터 전달 : 2D
odEncoder. fit(dataDF[['Grade' ]])
print(odEncoder.categories_)
print(odEncoder.n_features_in_)




## 스케일링 진행
encode_DF = odEncoder. transform(dataDF[ ['Grade']])

print(dataDF['Grade'].to_list())
print(encode_DF.tolist())
```
