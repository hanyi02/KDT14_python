```python
import numpy as np
```

```python
a = np.array([[1, 2],
              [3, 4]])
b = np.array([[5, 6],
              [7, 8]])
print(np.vstack([a, b]))
print(np.hstack([a, b]))
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
result = np.dstack((arr1, arr2))
print(result)
print(result.shape)
```

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.concatenate([a, b], axis=0))
print(np.concatenate([a, b], axis=1))
print(np.concatenate([a, b], axis=None))
```

```python
arr = np.arange(12).reshape(3, 4)
h_splits = np.hsplit(arr, 2)
print(h_splits[0])
print(h_splits[1])
```

```python
arr = np.arange(12).reshape(3, 4)
v_splits = np.vsplit(arr, 3)
print(v_splits[0])
print(v_splits[1])
print(v_splits[2])
```

```python
arr = np.arange(12).reshape(3, 4)
h_splits = np.hsplit(arr, [1, 3])
print(h_splits[0])
print(h_splits[1])
print(h_splits[2])
```

```python
arr = np.array([1, 2, 3, 4])
view_arr = arr.view()
copy_arr = arr.copy()
view_arr[0] = 100
copy_arr[1] = 200
print(arr)
print(view_arr)
print(copy_arr)
```

```python
x = np.array([1, 2, 3, 4, 5])
print(np.add.reduce(x))
print(np.multiply.reduce(x))
print(np.add.accumulate(x))
```

```python
arr = np.array([1, 2, 3, 4, 5, 6])
print(np.add.reduceat(arr, [0, 3]))
print(np.multiply.reduceat(arr, [0, 2, 4]))
```

```python
x = np.array([[1, 2, 3], [4, 5, 6]])
y = np.array([10, 20, 30])
print(np.add(x, y))

a = np.array([1, 2, 3])
b = np.array([2, 2, 2])
print(np.greater(a, b))
print(np.logical_and(a > 1, b < 3))
```
