### [ 정렬 & arg 함수명() ] <hr>

- 함수: np.sort()=> [기] 오름차순 정렬된 ndarray 반환
- 매서드: ndarray.sort()=> 원본 정렬
 
- arg000() 함수: 
    * 해당 값이 있는 인덱스 반환(값 반환 아님)
    * 예: ndarray.max(): 원소 중 가장 큰 값
    * 예: ndarray.argmax(): 원소 중 가장 큰 값의 인덱스

```python
# ---------------------------------------
# 모듈 로딩
# ---------------------------------------

import numpy as np
```

```python
# ---------------------------------------
# 실습 array 생성
# ---------------------------------------

a1= np.array([[3, 1, 6],
               [0, 7, 4],
               [5, 3, 2]])

print(a1)
```

```python
# ---------------------------------------
# 정렬
# ---------------------------------------
as1= np.sort(a1)

print("a1====>", a1.shape, a1, sep= '\n')
print("as1====>", as1.shape, as1, sep= '\n')

## 매서드: 원본이 바뀜
a1.sort()
print(a1)
```
