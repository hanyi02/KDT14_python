### [ 댜양한 ndarray 객체 생성 ] <hr>


```python
# 모듈 로딩
import numpy as np
```

```python
# --------------------------------------
#  다양한 타입의 데이터 => ndarray로 저장
#  사용 함수: np.array()
# --------------------------------------

arr1= np. array([11, 22, 33])
arr2= np. array([2.3, 0.9])
arr3= np. array(['A', 'B'])
arr4= np. array(['Good', 100, True]) # 서로 다른 타입의 데이터 경우 제일 큰 데이터 타입으로 일치시킴 배열은 동일 자료형만 다루기 때문

```

```python
# --------------------------------------
#  ndarray 객체 속성 출력 함수
#  함수 이름: print_info()
#  매개 변수: arr-- ndarray 객체
#            name-- ndarray 식별 문자열
#  반환값: 없음, 직접 출력
# --------------------------------------

def print_info(arr, name):
    print(f'{name}=> {arr}')
    print(f'{name}의 데이터 타입=> {arr.dtype}')
    print(f'{name}의 차원 수=> {arr.ndim}D 차원')
    print(f'{name}의 형태=> {arr.shape}')
    print(f'{name}의 원소 수=> {arr.size}개')
    print(f'{name}의 원소 크기=> {arr.itemsize} bytes')
    print(f'{name}의 value=> {arr}')
    print('-----------------------------')

```

```python
# --------------------------------------
#  ndarray 객체 속성 확인
# --------------------------------------
print_info(arr1, 'arr1')
print_info(arr2, 'arr2')
print_info(arr3, 'arr3')
print_info(arr4, 'arr4')
```
