### [ array 필터링 ] <hr>

- BooleanIndexing
- 함수: np.where( array[조건식] )=> 조건에 맞는 데이터/ 원소의 인덱스 튜플로 반환


```python
# ---------------------------------------
# 모듈 로딩
# ---------------------------------------

import numpy as np
```

```python
# ---------------------------------------
# 실습 array 생성
# ---------------------------------------

arr1d= np.arange(1, 31)

print(arr1d)
```

```python
# ---------------------------------------
# 필터링- (1) Boolean Indexing
# ---------------------------------------

bool_idx= arr1d%2==0

arr1d[bool_idx]
```

```python
# ---------------------------------------
# 필터링- (2) np.where(): 필터링 된 인덱스
# ---------------------------------------
filtering_idx= np.where(~arr1d%2)

arr1d[filtering_idx]
```

```python
# ---------------------------------------
# 필터링- (2) np.all(): 필터링 된 인덱스
# ---------------------------------------

arr2d= arr1d.reshape(-1, 10)

print("axis=None", np.all(arr1d%2), np.any(arr2d%2))
print("axis=1", np.all(arr2d%2, axis=1), np.any(arr2d%2, axis=1))
```
