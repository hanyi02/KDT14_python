### [ 축 변경 ] <hr>

- shape속성에 튜플 값의 순서를 변경
    * 2D shape은 (행, 열)
    * 3D shape은 (깊이, 행, 열)
    * 4D shape은 (개수, 깊이, 행, 열)
- 원소 개수 변경 X, 차원 변경 X
- 속성: ndarray.T
- 메서드: ndarray.transpose(*축들)/ ndarray.swapaxes(축1, 축2)

```python
# ---------------------------------------
# 모듈 로딩
# ---------------------------------------

import numpy as np
```

```python
# ---------------------------------------
# 실습 array 생성
# ---------------------------------------

a1= np.arange(1, 9).reshape(2, 4)
a2= np.arange(1, 5, 0.25 ).reshape(1, 4, -1)


print(a1.shape, a2.shape, a3.shape)
```

```python
# ---------------------------------------
# 행렬 내적 연산- (1) np.matmul()/ @
# ---------------------------------------
a12= np.matmul(a1, a2)
print("a12", a12, sep= '\n')

# np.matmul(a1, a3) ---> 행렬 식이 안 맞아서 에러뜸

a13= np.matmul(a1, a3.T) 
print("a13", a13, sep= '\n')
 

op= a1 @ a2
print("op====>", op, sep='\n')
```

```python
# ---------------------------------------
# 행렬 외적 연산- (2) np.outer()/ @
# ---------------------------------------
a12= np.outer(a1, a2)
print("a12", a12, sep= '\n')

# np.matmul(a1, a3) ---> 행렬 식이 안 맞아서 에러뜸

a13= np.outer(a1, a3.T) 
print("a13", a13, sep= '\n')
 
```
