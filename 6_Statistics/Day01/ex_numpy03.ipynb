### [ Numpy Fancy Indexing ] <hr>

- 원하는 데이터의 인덱스를 배열/ 리스트에 담아서 한꺼번에 추출하는 인덱싱
- 예) 배열명[[인덱스, 인덱스]]


```python
import numpy as np
```

```python
# 실습 데이터
dataNP= np.array([[11, 22, 33], [44, 55, 66]], dtype= np.int8)

print(dataNP)

# 팬시 인덱싱=> 2번, 0번 컬럼 추출
dataNP[:, [2, 0]]= 9

print(dataNP)
```

```python
# 팬시 인덱싱=> 1번, 0번 컬럼 추출
dataNP2= dataNP[[1, 0]]
print(dataNP2)
```
