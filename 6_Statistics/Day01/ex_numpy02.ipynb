## [ Numpy - ndarray 원소 접근 ] <hr>

- 인덱싱 기반: 왼-> 오, 오-> 왼 2개 종류 인덱스
- 슬라이싱: 연속적이거나 일정한 규칙이 있는 원소 접근 방식

```python
# 모듈 로딩
# ========================
import numpy as np
```

```python
# 실습 데이터 생성
data= np.array([1, 2, 3, 4, 5])
```

```python
print(data[0], type(data[0]), data[0].ndim) ## 0번 원소
print(data[-1], type(data[-1]), data[-1].ndim) ## 제일 마지막 원소
```

```python
## 원소값 변경
data[0] = 10
print(data)
```

```python
# ========================
# ndarray 원소 접근- 슬라이싱
# ========================

data[0:-2:2]
```

```python
# ========================
# 슬라이싱+ 값 변경
# ========================
# view 타입: 값 변경 시 원본 데이터 변경 됨
data[0:-3]= 7
print(data)

# copy 타입: 값 변경 시 원본 데이터 변경 안됨
data2 = data[0:-3].copy()
data2[0] = 100
print(data)

# 값 출력
print(data)
print(data2)
```
