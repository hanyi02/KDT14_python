- [ Ndarray 객체 생성 방법들 ] <hr>

```python
# 모듈 로딩
import numpy as np
```

```python
# [방법 1] np.array() 함수 이용
# ========================================
arr1= np.array([11, 22, 33])

print(arr1) #리스트랑은 다름. 리스트는 쉼표가 있고, array는 쉼표가 없다.
print(f'shape: {arr1.shape}')
print(f'ndim: {arr1.ndim} D')
print(f'dtype: {arr1.dtype}')
print(f'itemsize: {arr1.itemsize} bytes')
print(f'data: {arr1.data}') #메모리 주소값이 나온다. array는 연속된 메모리 공간에 저장되기 때문에 주소값이 나온다.
```

```python
# [방법 1] np.array() 함수 이용
# ========================================
arr1= np.array([[11, 22, 33]]) # 괄호가 2개= 2차원 배열

print(arr1) #리스트랑은 다름. 리스트는 쉼표가 있고, array는 쉼표가 없다.
print(f'shape: {arr1.shape}')
print(f'ndim: {arr1.ndim} D')
print(f'dtype: {arr1.dtype}')
print(f'itemsize: {arr1.itemsize} bytes')
print(f'data: {arr1.data}') #메모리 주소값이 나온다. array는 연속된 메모리 공간에 저장되기 때문에 주소값이 나온다.
print(f'strides: {arr1.strides}') # (행 이동을 위한 메모리칸수, 열 이동을 위한 메모리칸수)
```

```python
 
arr1= np.array([[11, 22, 33], [44, 55, 66]], dtype=np.int16) # dtype을 지정해주면, 데이터 타입이 바뀐다. (기본값은 int64)

print(arr1) #리스트랑은 다름. 리스트는 쉼표가 있고, array는 쉼표가 없다.
print(f'shape: {arr1.shape}')
print(f'ndim: {arr1.ndim} D')
print(f'dtype: {arr1.dtype}')
print(f'itemsize: {arr1.itemsize} bytes')
print(f'data: {arr1.data}') #메모리 주소값이 나온다. array는 연속된 메모리 공간에 저장되기 때문에 주소값이 나온다.
print(f'strides: {arr1.strides}') # (행 이동을 위한 메모리칸수, 열 이동을 위한 메모리칸수)
```

```python
# ========================================
# [방법 2] 특정값으로 채운 ndarray 생성
# ========================================
# 1) np.zeros() 함수 이용
arr1=np.zeros( (3,) ) # 3행 4열 2차원
arr2=np.zeros((2, 4)) # 1차원
arr3=np.zeros( (2, 4), dtype=np.int8 )

print(arr1)
print(arr2)
print(arr3)
```

```python
# np.ones() 함수 이용
arr1=np.ones( (3) ) 
arr2=np.ones((3, 2))
arr3=np.ones((3, 2), dtype=np.bool)

print(arr1)
print(arr2)
print(arr3)
```

```python
# ========================================
# [방법 3] 범위 지정해서 ndarray 생성
# ========================================
# => np.arange() 함수 이용
arr1= np.arange(5)

# => np.linspace() 함수 이용
arr2, step=np.linspace(0, 5, retstep=True) # 0부터 5까지 1씩 증가하는 숫자들을 생성한다. (endpoint=False로 하면, 5는 포함되지 않는다.)

# => np.logspace() 함수 이용
arr3=np.logspace(0, 5, num=10) # 10개의 숫자들을 생성한다. (10^0부터 10^5까지)

print("arr1", arr1, sep="\n")
print("arr2", step, arr2, sep="\n")
print("arr3", arr3, sep="\n")

```

```python
# np.full(): 원하는 특정 값으로 채운 ndarray 생성
arr1=np.full((2, 3), fill_value=7) # 2행 3열의 배열을 생성하고, 모든 원소를 7로 채운다.
print("arr1", arr1, sep="\n")
```

```python
# ========================================
# [방법 4] 특정 형태의 배열 객체 생성
# ========================================

np.empty((3, 2)) # 2행 3열의 배열을 생성하고, 모든 원소를 0으로 채운다. (초기화되지 않은 배열을 생성한다.)

# => np.empty_like() 함수 이용
arr2=np.empty_like(arr1) # arr1과 같은 형태의 배열을 생성한다. (초기화되지 않은 배열을 생성한다.)

arr2
```
