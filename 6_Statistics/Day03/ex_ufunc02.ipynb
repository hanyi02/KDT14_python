### [ 유니버셜 함수- Numpy 성능 핵심 요소 ] <hr>
- 반복문 없이 원소 단위 연산 수행하는 함수
- 타입과 형태 변환 최적화 진행=> 브로드 캐스팅 자동 진행


```python
import numpy as np
```

```python
# 실습 데이터 생성
arr= np.array([3.2, 0.1, 8., -5.2, -3.12])
print(arr.dtype, arr.itemsize, arr.size,  sep=',')
print(arr)
```

```python
# --------------------------------------
# 실수 데이터=> 소수점 이하 자릿수 관련 함수
# --------------------------------------
# => np.ceil(): 소수점 이하는 모두 올림 처리
print("원본: ", arr)

retarr= np.ceil(arr)
for idx in range(arr.size):
    print("반환결과=> ", arr[idx], retarr[idx])


```

```python
# => np.round(): 소수점 이하 5이상 반올림 처리
print("원본: ", arr)

retarr= np.round(arr, 1)
for idx in range(arr.size):
    print("반환결과=> ", arr[idx], retarr[idx])
```

```python
# => np.floor(): 소수점 이하 내림 처리
print("원본: ", arr)

retarr= np.floor(arr)
for idx in range(arr.size):
    print("반환결과=> ", arr[idx], retarr[idx])
```

```python
# -----------------------------------
# 누적 연산 처리 함수
# -----------------------------------
# => 연산.reduce(array_obj): 누적 연산 결과 반환 
arr= np.arange(1, 10, 3)
result= np.add.reduce(arr)
print(f'{arr}=> {result}')

result= np.multiply.reduce(arr)
print(f'{arr}=> {result}')
```

```python
# => 연산.accumulate(array_obj): 누적 연산 값 반환 


result= np.add.accumulate(arr)
print(f'{arr}=> {result}')
```

```python
# -----------------------------------
# 누적 연산 처리 함수 - 2D ndarray
# -----------------------------------

arr= np.arange(1, 17).reshape(4, 4)

# => 2개 컬럼 묶어서 연산
retarr= np.add.reduceat(arr, [0, 2], axis=1)

print(f'{arr}\n ---연산 후---\n {retarr}')
```
