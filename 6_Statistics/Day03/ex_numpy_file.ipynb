### [ Numpy 파일 입출력 ] <hr>
- ndarray 객체를 파일로 저장
- 확장자: 1개의 파일-> 파일명.npy
    *  여러개의 파일-> 파일명.npz
    


```python
import numpy as np
```

```python
# 실습 데이터 생성
a1= np.linspace(1, 3)
a1
```

```python
# --------------------------------------
# ndarray 저장
# --------------------------------------

data_file= './data.npy'
all_file= './all.npz'
com_file='./com.npz'
text_file= './text.out'

# 1개 ndarray=> 파일로 저장
np.save(data_file, a1)

# 여러개 ndarray=> 파일로 저장
np.savez(all_file, a1, a1, a1)

# 여러개 ndarray=> 압축 후 파일로 저장
np.savez_compressed(com_file, x=a1, y=a1, z=a1)

# 여러개 ndarray=> 텍스트 파일로 저장
np.savetxt(text_file, a1)
```

```python
# --------------------------------------
# ndarray 데이터 파일 읽기
# --------------------------------------
# 파일 1개 읽기
np.load(data_file)

# 파일 여러개 읽기
ret= np.load(all_file)
print(f'ret=> {type(ret)}')
print(f'files=> {ret.files}')
print(f'keys=> {ret.keys}')


# 텍스트로 저장된 파일 읽기
ret= np.loadtxt('text.out')
ret
```

```python
# -----------------------------------
# 누적 연산 처리 함수
# -----------------------------------
# => 연산.reduce(array_obj): 누적 연산 결과 반환 
arr= np.arange(1, 10, 3)
result= np.add.reduce(arr)
print(f'{arr}=> {result}')

result= np.multiply.reduce(arr)
print(f'{arr}=> {result}')
```

```python
# => 연산.accumulate(array_obj): 누적 연산 값 반환 


result= np.add.accumulate(arr)
print(f'{arr}=> {result}')
```

```python
# -----------------------------------
# 누적 연산 처리 함수 - 2D ndarray
# -----------------------------------

arr= np.arange(1, 17).reshape(4, 4)

# => 2개 컬럼 묶어서 연산
retarr= np.add.reduceat(arr, [0, 2], axis=1)

print(f'{arr}\n ---연산 후---\n {retarr}')
```
