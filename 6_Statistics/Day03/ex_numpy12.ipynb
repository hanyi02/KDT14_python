### [ ndarray 이미지 데이터 ] <hr>
- python에서 이미지 파일 로딩==> ndarray 타입
- 화면 그림 출력: matplotlib.plot.imshow()
- 컴비전: opencv 패키지- numpy, pillow 패키지- tensor, DL

```python
# 모듈 활성화
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.cm as cm
```

```python
# 실습 ndarray 생성
data= np.arange(0, 256).reshape((16, 16))

data.size, data.shape
```

```python
plt.imshow(data, cmap=cm.gray)
plt.show()

```
